```python
import ollama
```

```python
def generate_text(prompt):
    res=ollama.generate(model='llama3.2:1b',prompt=prompt)
    return res['response']
```

```python
def answer_question(question):
    prompt=f'please write a sentence about {question}'
    # res=ollama.generate(model='llama3.2:1b',prompt=prompt)
    # return res['response']
    res=generate_text(prompt)
    return res
```

```python
def summurize_text(text):
    prompt=f'please summurize the text given in one sentence {text}'
    res=generate_text(prompt)
    return res
```

```python
def translate_text(text,language):
    prompt=f'please translate the given text {text} to {language}'
    res=generate_text(prompt)
    return res
```

```python
print("OLLAMA TEXT BASED TASKS")
print("*"*50)
print('1.Text Generation')
print('2.Answer Your Question')
print('3.Summerize The Text')
print('4.Translate The Text')
choice=int(input("enter your choice:"))
if choice==1:
    prompt=input('enter the text:')
    result=generate_text(prompt)
elif choice==2:
    question=input('state your question:')
    result=answer_question(question)
elif choice==3:
    summerize=input('enter the topic to summurize:')
    result=summurize_text(summerize)
elif choice==4:
    text=input('enter the text:')
    lang=input('enter the language:')
    result=translate_text(text,lang)
else:
    result='wrong input'
print('-'*50,'RESULT','-'*50)
print()
print(result)
```
